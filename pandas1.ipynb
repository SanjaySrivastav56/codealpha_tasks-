```python
import pandas as pd
import numpy as np

# Define the data dictionary
data = {
    "name" : ["sanjay","srivastav","Lakshman","Arjun"],
    "Age" : [20,30,35,18],
    "department" : ["IT","Finance","Charted Accountant","Bank Employee"],
    "Salary" : [10000,20000,30000,40000]
}

# Create the DataFrame
my_df = pd.DataFrame(data)

# Renaming columns
my_df.rename(columns = {"Salary": "salary"}, inplace = True)
my_df.rename(columns = {"Age":"age"}, inplace = True)

# Creation of new column
my_df["promoted salary"] = my_df["salary"] * 10

# Modifying in series by adding a new row
my_df.loc[4] = ['Aarav', np.nan,"IT",np.nan,np.nan]

#duplicating
my_df.dup = my_df[my_df.duplicated()]
```

```python
my_df
```

```python
duplicate = my_df[my_df.duplicated()]
duplicate
```

```python
#if we keep = "first" it means the last row is duplicated
duplicate = my_df[my_df.duplicated(keep = "first")]
duplicate
```

```python
#if we keep = "last" it means the first row is duplicated, you can observe by seeing index,value is changed
duplicate = my_df[my_df.duplicated(keep = "last")]
duplicate
```

```python
dropping_duplicates = my_df.drop_duplicates()
dropping_duplicates
```

```python
my_df
```

```python
 #invslid values
 #lambda functions -> python
 #here x is number of records
my_df["promoted salary"] = my_df["promoted salary"].apply(lambda x: x/6 if x > 50000 else x)
my_df
```

```python
#apply and lambda
def multiply_age(x):
  return x*2

my_df["age"] = my_df["age"].apply(multiply_age)
my_df
```

```python
my_df["age"] = my_df["age"].apply(lambda x: x/2)
my_df
```

```python
#joins and merge
department_info = {
    "department" : ["HR","IT","Finance"],
    "Location" : ["New York","San Francisco","Chicago"],
    "Manager":["Laura","Steve","Nina"]
}
df2 = pd.DataFrame(department_info)
df2
```

```python
#combine dataset using concat()
#this is example of row wise
pd.concat([my_df,df2])
```

```python
#concat by columns it will match department and give loc,manager to the same name by identifying same dept
pd.concat([my_df,df2],axis = 1)
```

```python
my_df.dropna(how = "any" , inplace = True)
```

```python
#only one department column is reflected it will automatically match the department and location,manager are same
pd.merge(my_df,df2, on = "department")
```
